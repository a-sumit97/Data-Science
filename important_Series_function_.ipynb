```python
#drop duplicates

temp = pd.Series([1,1,2,3,4,4,3,5,5,5])
temp.drop_duplicates()
```

```python
#check duplicates

temp[temp.duplicated()]
```

```python
#isnull

temp = pd.Series([1,2,3,np.nan,4,np.nan])
temp[temp.isnull()]
```

```python
#count
temp.count() #count only not null values
```

```python
#size
temp.size # total number of entries including nan
```

```python
vk.isnull().sum()
```

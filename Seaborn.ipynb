### 1. Relational Plot

- to see the statistical relation between 2 or more variables.
- Bivariate Analysis

Plots under this section

- scatterplot
- lineplot

image.png

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
# scatter plot -> axis level -> rectangle shape

sns.scatterplot(x='total_bill', y='tip', data=tips, hue = 'sex', style = 'time', size = 'size' )
```

```python
# relplot -> figure level -> square shape

sns.relplot(x='total_bill', y='tip', data=tips, hue = 'sex', style = 'time', size = 'size', kind = 'scatter')
```

```python
# line plot

gap = px.data.gapminder()
temp_df = gap[(gap['country'] == 'India')]
```

```python
# axes level function

sns.lineplot(x = 'year', y='lifeExp', data = temp_df)
```

```python
# figure level

sns.relplot(x = 'year', y='lifeExp', data = temp_df, kind='line', color = 'red')
```

```python
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(x = 'year', y='lifeExp', data = temp_df, hue = 'country', kind = 'line', size = 'continent')
```

```python
sns.lineplot(x = 'year', y='lifeExp', data = temp_df, hue = 'country', style = 'continent')
```

```python
#facet plot - > figure level function -> work with relplot
# it will not work will scatterplot and lineplot

sns.relplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', col = 'sex')
```

```python
sns.relplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', row = 'sex')
```

```python
sns.relplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex',col ='smoker', row = 'sex', size = 'size', kind= 'scatter')
```

```python
sns.relplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex',col ='smoker', row = 'day')
```

```python
sns.scatterplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', col = 'sex')
```

```python
# col wrap
sns.relplot(x = 'lifeExp', y = 'gdpPercap', kind = 'scatter', data = gap, col='year',col_wrap = 3)
```

# Histogram

image.png

```python
# Figure level - displot
# Axes level - histplot, kdeplot, rugplot
```

```python
# plotting univariate histogram

tips
```

```python
#figure level
sns.displot(data = tips, x = 'total_bill',hue='sex',element = 'step')
```

```python
#axes level

sns.histplot(data = tips, x = 'total_bill',hue='sex',element = 'step')
```

```python
#facetplot using col and row -> not work on histplot function

sns.displot(x='total_bill',data = tips, col='sex',element = 'step',row='time')
```

```python
# kdeplot
# Rather than using discrete bins, a KDE plots smooths the obsevations with a Gaussian Kernel, producing a continuous destiny estimate
```

```python
sns.kdeplot(x='total_bill',data=tips)
```

```python
sns.displot(x='total_bill',data=tips,kind='kde')
```

```python
from textwrap import fill
#hue->fill
sns.displot(x='total_bill',data=tips,kind='kde',hue='sex')
```

```python
sns.displot(x='total_bill',data=tips,kind='kde',hue='sex',fill=True,height=10,aspect=2)
```

```python
# Rugplot

# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')

sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observations within each rectangle with the fill color

# sns.histplot(data=tips, x='total_bill', y='tip')
sns.displot(data=tips, x='total_bill', y='tip',kind='hist')
```

```python
sns.histplot(data=tips, x='total_bill', y='tip')
```

```python
#bivariate kdeplot
# a bivariate kdeplot smoothes the (X,y) obeservations within a 2d gaussian

sns.kdeplot(x='total_bill',y='tip',data=tips)
```

```python
sns.displot(x='total_bill',y='tip',data=tips,kind='kde')
```

### 2. Matrix Plot

- Heatmap
- Clustermap


```python
# Heatmap

# Plot rectangular data as a color-encoded matrix
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')
temp_df
```

```python
#axes level funtion
plt.figure(figsize=(15,10))
sns.heatmap(temp_df)
```

```python
#annot
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,10))
sns.heatmap(temp_df,annot=True)
```

```python
##linewidth
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,10))
sns.heatmap(temp_df,annot=True,linewidth=0.5)
```

```python
##cmap
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,10))
sns.heatmap(temp_df,annot=True,linewidth=0.5,cmap='viridis')
```

```python
# Clustermap

# Plot a matrix dataset as a hierarchically-clustered heatmap.

# This function requires scipy to be available.

iris = px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```

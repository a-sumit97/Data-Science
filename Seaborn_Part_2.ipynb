### Figure level function -> `catplot`

```python
#Strip plot
# Axes level function
sns.stripplot(x='day', y='total_bill', data=tips)
```

```python
sns.stripplot(x='day', y='total_bill', data=tips,jitter=False)
```

```python
#jitter
sns.catplot(x='day',y='total_bill',data=tips,kind='strip',jitter=0.2,hue='sex')
```

```python
#swarmplot

sns.swarmplot(x='day',y='total_bill',data=tips)
```

```python
sns.catplot(x='day',y='total_bill',data=tips,kind='swarm',hue='sex')
```

### Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

image.png

```python
#Box plot

sns.boxplot(data=tips,x='sex',y='total_bill')
```

```python
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
#using catplot

sns.catplot(data=tips,x='sex',y='total_bill',kind='box')
```

```python
# hue
sns.catplot(data=tips,x='day',y='total_bill',kind='box',hue='sex')
```

```python
# single boxplot -> numerical col

sns.catplot(data=tips,y='total_bill',kind='box')
```

# Violinplot = (Boxplot + KDEplot)

```python
sns.catplot(x='day',y='total_bill',data=tips,kind='violin')
```

When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
sns.barplot(x='sex',y='total_bill',data=tips,hue='smoker')
```

```python
sns.catplot(x='sex',y='total_bill',data=tips,hue='smoker',kind='bar')
```

```python
import numpy as np
sns.catplot(x='sex',y='total_bill',data=tips,hue='smoker',kind='bar',estimator=np.std)
```

```python
import numpy as np
sns.catplot(x='sex',y='total_bill',data=tips,hue='smoker',kind='bar',estimator=np.min)
```

```python
#pointplot

import numpy as np
sns.catplot(x='sex',y='total_bill',data=tips,hue='smoker',kind='point',estimator=np.std)
```

```python
sns.pointplot(x='sex',y='total_bill',data=tips,hue='smoker',errorbar=None)
```

A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable

```python
#countplot

sns.countplot(data=tips,x='sex',hue='day')
```

```python
#faceting using catplot

sns.catplot(data=tips,x='sex',y='total_bill',col="smoker",kind='box',row='time')
```

### Regression Plots

- regplot
- lmplot

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
#axes level
#hue parameter is not available
sns.regplot(data=tips,x='total_bill',y='tip')
```

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
```

```python
#residplot
sns.residplot(x='total_bill',y='tip',data=tips)
```

A second way to plot Facetplot -> FacetGrids

```python
#figure level -> relplot -> displot -> catplot -> lmplot

sns.catplot(data=tips,x='sex',kind='violin',y='total_bill',col='day',row='time')
```

```python
g = sns.FacetGrid(data=tips,col='day',row='time')
g.map(sns.violinplot,'sex','total_bill')
```

# Plotting Pairwise Relationship (Pairgrid vs Pairplot)

```python
iris
```

```python
sns.pairplot(iris)
```

```python
sns.pairplot(iris,hue='species')
```

```python
# Pair grid

g = sns.PairGrid(iris,hue='species')
g.map(sns.scatterplot)
```

```python
# Map_diag -> map_offdiag

g = sns.PairGrid(iris,hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
g = sns.PairGrid(iris,hue='species')
g.map_diag(sns.violinplot)
g.map_offdiag(sns.scatterplot)
```

```python
#map_upper-> map_lower
g = sns.PairGrid(iris,hue='species')
g.map_diag(sns.violinplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.histplot)
```

```python
 #map_upper-> map_lower
g = sns.PairGrid(iris,hue='species',vars=['sepal_length','sepal_width'])
g.map_diag(sns.violinplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.histplot)
```

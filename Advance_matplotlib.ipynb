```python
iris = pd.read_csv('/content/iris.csv')
iris.sample(5)
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-virginica':1,'Iris-versicolor':2})
iris.sample(5)
```

```python
plt.figure(figsize=(13,7))

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap = 'summer',alpha = 0.3)
plt.title('Scatterplot')
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')

plt.colorbar()

plt.show()
```

# 3d Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy = np.meshgrid(x,y)

xx.shape
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize = (15,10))

ax = plt.subplot(projection = '3d')

p = ax.plot_surface(xx,yy,z,cmap = 'viridis')

fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
z.shape
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize = (15,10))
ax = plt.subplot(projection = '3d')

p = ax.plot_surface(xx,yy,z,cmap = 'viridis')

fig.colorbar(p)
```

```python
z = np.sin(xx) + np.log(yy)

fig = plt.figure(figsize = (15,10))
ax = plt.subplot(projection = '3d')

p = ax.plot_surface(xx,yy,z,cmap = 'viridis')
```

```python
delivery = pd.read_csv('/content/IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run']==6)]
```

```python
grid = temp_df.pivot_table(index = 'overs', columns = 'ballnumber', values = 'batsman_run', aggfunc = 'count')
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.yticks(delivery['overs'].unique(),list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.colorbar()
```

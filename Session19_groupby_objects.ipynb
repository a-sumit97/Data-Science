```python
# find number of movies starting with A for each group

def foo(group):
  return group['Series_Title'].str.startswith('A').sum()
```

```python
genre.apply(foo)
```

```python
# find ranking of each movie in the group according to IMDB rating

def rank_movies(group):
  group['Genre_rank'] = group['IMDB_Rating'].rank(ascending = False)
  return group
```

```python
genre.apply(rank_movies)
```

```python
#find normalized IMDB rating group wise

def normalized_group(group):
  group['normalized_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group
```

```python
genre.apply(normalized_group)
```

```python
ipl = pd.read_csv('/content/deliveries.csv')
ipl
```

```python
ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(5)
```

```python
six = ipl[ipl['total_runs'] == 6]
six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(5)
```

```python
#find batsman with most number of 4 and 6 in last 5 over
temp_df = ipl[ipl['over'] > 15]
temp_df = temp_df[(temp_df['batsman_runs'] ==  6) | (temp_df['batsman_runs'] == 4)]
temp_df.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(5)
```

```python
#find v kohli's record against all team

vk = ipl[ipl['batsman'] == 'V Kohli']
vk.groupby('bowling_team')['batsman_runs'].sum()
```

```python
ipl.sample(5)
```

```python
#create a function that can return the highest runs of a batsman

temp_df = ipl[ipl['batsman'] == 'V Kohli']
temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
def highest_runs(batsman):
  temp_df = ipl[ipl['batsman'] == batsman]
  return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
highest_runs('MS Dhoni')
```
